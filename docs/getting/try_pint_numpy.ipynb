```python
import numpy as np  # noqa: F401

import pint

ureg = pint.UnitRegistry()
Q_ = ureg.Quantity
```

Pint lets you attach units to values and lets Python do the unit math for you:

```python
distance = Q_("42 km")
print(distance.to("mile"))
```

Pint checks dimensionality for you:

```python
try:
    distance.to("joule")
except pint.DimensionalityError as exc:
    print(exc)
```
