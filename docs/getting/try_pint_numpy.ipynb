```python
import numpy as np

import pint

ureg = pint.UnitRegistry()
Q_ = ureg.Quantity
```

A `Quantity` can wrap a NumPy array just like it wraps a plain number:

```python
legs1 = Q_(np.asarray([3.0, 4.0]), "meter")
legs1
```

Unit conversion works the same way as for a scalar `Quantity`:

```python
legs1.to("kilometer")
```

Pint checks dimensionality for you, even on array operations:

```python
try:
    legs1.to("joule")
except pint.DimensionalityError as exc:
    print(exc)
```

Many NumPy functions are unit-aware. For example, `np.hypot` converts its
arguments to consistent units before combining them, and `np.arccos` expects
a dimensionless ratio and returns an angle:

```python
legs2 = Q_([400.0, 300.0], "centimeter")
hyps = np.hypot(legs1, legs2)
hyps
```

```python
angles = np.arccos(legs2 / hyps)
angles.to("degree")
```
